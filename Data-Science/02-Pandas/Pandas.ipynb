# Series
A Pandas 1-Dimensional labeled array that can hold any data type. Think of it like a single column in a spreadsheet (1-Dimensional)

```python
data = [100, 102, 104, 200, 202]
series = pd.Series(data, index=["a", "b", "c", "d", "e"])
print(series)
```

Accessing value directly in series

```python
print(series.loc["a"])
```

For changing the value by calling the index

```python
# series.loc["c"] = 200
# print(series)
```

```python
print(series.iloc[2])
```

```python
print(series)
```

Filtering series

```python
print(series[series >= 200])
print(series[series < 200])
```

Using Dictionaries

```python
calories = {"Day 1" : 1750, "Day 2": 2100, "Day 3" : 1700}
```

```python
series = pd.Series(calories)
print(series)
```

# DataFrame
- A tabular data structure with rows AND columns. (2 Dimensional)
- Similar to an Excel spreadsheet.

```python
data = {"Name" : ["Spongebob", "Patrick", "Squidward"],
        "Age" : [30, 35, 50]
        }
```

```python
df = pd.DataFrame(data, index=["Employee 1", "Employee 2", "Employee 3"])
print(df)
```

Accessing by index

```python
print(df.loc["Employee 2"])
print(df.iloc[2])
```

```python
print(df)
```

## Adding new Column

```python
df["Job"] = ["Cook", "N/A", "Cashier"]
print(df)
```

## Adding new Row

```python
new_rows = pd.DataFrame([{"Name" : "Sandy", "Age" : 28, "Job" : "Engineer"},
                         {"Name" : "Eugene", "Age" : 60, "Job" : "Manager"}], 
                        index=["Employee 4", "Employee 5"])
df = pd.concat([df, new_rows])
print(df)
```

## Filtering
- Keeping the rows that Match a <strong>condition<strong>

```python
df = pd.read_csv("data.csv")
tall_pokemon = df[df["Height"] >=2 ]

print(tall_pokemon)
```

```python
df = pd.read_csv("data.csv")
heavy_pokemon = df[df["Weight"] > 100 ]

print(heavy_pokemon)
```

```python
df = pd.read_csv("data.csv")
legend_pokemon = df[df["Legendary"] == 1 ]

print(legend_pokemon)
```

```python
df = pd.read_csv("data.csv")
water_pokemon = df[(df["Type1"] =="Water") | (df["Type2"] == "Water")]

print(water_pokemon)
```

```python
df = pd.read_csv("data.csv")
ff_pokemon = df[(df["Type1"] == "Fire") & (df["Type2"] == "Flying")]

print(ff_pokemon)
```

## Aggregation
- Reduces a set of values into a single summary value
- Used to summarize and analyze data
- Often used with the groupby() funtion

### Aggregate Functions on complete dataset

### Mean

```python
print(df.mean(numeric_only=True))
```

### Sum

```python
print(df.sum(numeric_only=True))
```

### Minimum

```python
print(df.min(numeric_only=True))
```

### Max

```python
print(df.max(numeric_only=True))
```

### Count

```python
print(df.count())
```

### Aggregate Functions on single column

```python
print(df["Height"].mean())
print(df["Height"].sum())
print(df["Height"].min())
print(df["Height"].max())
print(df["Height"].count())
```

## <strong>Grouping<strong>

### <strong>Group by Object<strong>

# <strong>Data Cleaning<strong>
- The process of removing/fixing incomplete, incorrect, or irrelevant data.
- ~75% of work donw with Pandas is data cleaning.

1. Drop irrelevant columns

```python
df2 = df.drop(columns=["Legendary", "No"])
print(df2)
```

2. Handle missing data

```python
df3 = df.dropna(subset=["Type2"])
print(df3.to_string())
```

```python
df4 = df.fillna({"Type2": "None"})
print(df4.to_string())
```

Fixing inconsistent values

```python
df["Type1"]=df["Type1"].replace({"Grass": "GRASS",
                                 "Fire" : "FIRE",
                                 "Water" : "WATER"})

print(df.to_string())
```

Standardizing text

```python
df["Name"] = df["Name"].str.lower()
print(df.to_string())
```

Fixing DataTypes

```python
df["Legendary"] = df["Legendary"].astype(bool)
print(df.to_string())
```

Remove Duplicate Values

```python
df = df.drop_duplicates()
print(df)
```
